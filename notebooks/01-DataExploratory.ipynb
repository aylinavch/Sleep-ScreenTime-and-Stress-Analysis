# Análisis exploratorio: Sleep, Screen Time and Stress

Dataset: [Sleep, Screen Time and Stress Analysis](https://www.kaggle.com/datasets/jayjoshi37/sleep-screen-time-and-stress-analysis)

## 1. Carga de datos

```python
import pandas as pd

DATA_PATH = "../data/raw/sleep_mobile_stress_dataset_15000.csv"

df = pd.read_csv(DATA_PATH)
df.head()
```

## 2. Estructura y tipos de datos

```python
print(f"Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
df.info()
```

## 3. Valores nulos y duplicados

```python
print("Valores nulos por columna:")
print(df.isnull().sum())
print(f"\nFilas duplicadas: {df.duplicated().sum()}")
```

## 4. Estadísticas descriptivas

```python
df.describe()
```

```python
df.describe(include="object")
```

## 5. Distribución de variables categóricas

```python
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

## 6. Correlación entre variables numéricas

Correlación con `stress_level` para identificar posibles relaciones.

```python
numeric_cols = df.select_dtypes(include="number").columns
corr = df[numeric_cols].corr()
corr["stress_level"].sort_values(ascending=False)
```

## 7. Promedios agrupados por ocupación

```python
df.groupby("occupation")[["sleep_duration_hours", "daily_screen_time_hours", "stress_level"]].mean().sort_values("stress_level", ascending=False)
```
